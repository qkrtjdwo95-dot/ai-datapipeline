### python 리스트(배열) 과 NumPy 배열 비교

- 파이썬에 기존 리스트(배열) 자료구조 존재.속도전면에서 NumPy가 우세

#### 파이썬으로 배열작업

```python
prices = [10000, 15000, 12000, 30000, 25000]
discount_prices = []

for price in prices:
     discount_prices.append(price*0.9) # 10% 할인

discount_prices #노트북 코드셀에서 하나만 출력하려면 print() 생략 가능     
```

#### 넘파이로 배열작업

```python
np_prices = np.array(prices) # 넘파이 배열로 변경
np_discount_prices = np_prices*0.9

np_discount_prices. tolist()
```

#### NumPy 데이터타입
- int8 : -128 ~ 127 (합 256)
- int16 : -32768 ~ 32767(chd 65536)
- int32 : 32비트 정수
- int64 : 64비트 정수
- unit8 : 0~255
- unit16 : 0~65535
- float16 : 실수
- float32 : 32비트 실수
- float64 : 64비트 실수 (Numpy 기본 실수)
- complex64 : 복소수
- complex128 : 128비트 복소수
- bool : True/False
- str_ : 문자열
- object_ : 객체

#### 조건 필터링
- SQL WHERE절과 동일, 조건에 맞는 데이터만 골라내기

```python
np_prices
```

```python
np_prices >= 20000
```

```python
condition = np_prices >=20000
condition
```

```python
np_prices[condition]
```

```python
datas = np.array([-10, -2, 80, 999, 30, 120])
datas
```

```python
datas[(datas >= 0) & (datas <= 100)]
```
