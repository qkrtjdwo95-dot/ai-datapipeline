- 결측치 정리방법 : 결측치를 제거하거나(건수가 줄어듬), 나머지 평균값을 임의로 기입하거나, 기본통계에서 도출할 수 있는 다른 값들로 채우기

#### 이상치 핸들링
- 일반적인 범위를 벗어나는 비정상 데이터
- 예로 부산 1월 일별 평균기온에 -90도. 습도99%

```python
temperatures = np.array([22.5, 23.0, np.nan, 25.1, 99.0, 21.8, -99.0])
temperatures
```

```python
valid_temps = temperatures[
    (~np.isnan(temperatures))&  # nan은 무조건 제거
    (temperatures >= 15.0) &    # 최저기온이 15 아래로 내려갈 수 없으므로 제거
    (temperatures <= 50)        # 최고기온이 50이상 오를 수 없으므로 제거
]

valid_temps
```

- 전처리(PreProcessing) - 분석 이전에 이상치, 결측치를 없애는 작업
